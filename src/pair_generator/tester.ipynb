# Pair-selection interface prototype

Widgets for configuring a pair-screening window, sectors/regions, and pair count. The database queries and correlation/cointegration scoring are unfinished; the generation button is disabled.

Saved widget outputs show the original interface. To try the controls, run the notebook with `ipywidgets` installed. See the [database guide](../database/README.md) for the storage layer.


```python
import ipywidgets as widgets
from IPython.display import display, clear_output
from enum import Enum

class FormTag(Enum):
    """
    Enum for the form tags
    """
    UPDATE_FREQ = 0
    PC_WINDOW = 1
    PC_CORRELWEIGHT = 2
    REPEAT_COMPANIES = 3
    SECTORS = 4
    REGIONS = 5
    PAIRS_SCOPE = 6
    QUANTILE = 7
    NUM_PAIRS_VALUE = 8
    
Form_Holder = {
    FormTag.UPDATE_FREQ :None,
    FormTag.PC_WINDOW :None,
    FormTag.PC_CORRELWEIGHT :None,
    FormTag.REPEAT_COMPANIES :None,
    FormTag.SECTORS :None,
    FormTag.REGIONS :None,
    FormTag.PAIRS_SCOPE :None,
    FormTag.NUM_PAIRS_VALUE :None,
    FormTag.QUANTILE :None,
}

output = widgets.Output()
 
update_freq = widgets.ToggleButtons(
    options=['Monthly', 'Yearly'],
    description='Update frequency:',
    button_style='',
    tooltips=['More accurate, but slower', 'Quicker, but less accurate']
)
 
pc_window = widgets.SelectionSlider(
    options=[3, 6, 12, 24, 36, 60, 120, 240],
    value=12,
    description='Window of relevance for the Pair Coefficient (months):',
    orientation='horizontal',
    readout=True,
    readout_format='d',
    style={'description_width': 'auto'},
    layout=widgets.Layout(width='500px'), 
)
 
pc_correlweight = widgets.FloatSlider(
    value=0.50,
    min=0.00,
    max=1.00,
    step=0.10,
    description='Correlation Weight in Pair Coefficient*',
    orientation='horizontal',
    readout=True,
    readout_format='.2f',
    style={'description_width': 'auto'},
    layout=widgets.Layout(width='500px'), 
)

pc_correlweight_info = widgets.Label(
    value='*The Pair Coefficient is a weighted average of the correlation and cointegration',
    layout=widgets.Layout(width='500px'), 
)

repeat_companies = widgets.ToggleButtons(
    options=['Yes', 'No'],
    description='Companies are allowed to be in more than one pair:',
    button_style='',
    tooltips=['Better pairs, but less diversification', 'More diversification, but worse pairs']
)
 
sectors = widgets.SelectMultiple(
    options=['Energy','Materials','Industrials','Utilities', 'Healthcare', 'Financials', 'Consumer Discretionary',
    'Consumer Staples','Information Technology','Communication Services','Real Estate'],
    value=['Energy'],
    description='GICS sectors of interest:',
    style={'description_width': 'auto'},
    layout=widgets.Layout(width='500px'), 
)
 
but_allsectors = widgets.Button(description = 'Select all sectors')
but_allsectors_toggle = True
def select_allsectors(*args):
    global but_allsectors_toggle
    if but_allsectors_toggle:
        sectors.value = tuple(sectors.options)
        but_allsectors.description = 'Unselect all sectors'
        but_allsectors_toggle = False
    else:
        sectors.value = []
        but_allsectors.description = 'Select all sectors'
        but_allsectors_toggle = True
but_allsectors.on_click(select_allsectors)
 
regions = widgets.SelectMultiple(
    # TODO change to countries in the DB (can do it manually)
    options=['North America','Latin America','Europe','EMEA','APAC'],
    value=['North America'],
    description='Regions of interest:',
    style={'description_width': 'auto'},
    layout=widgets.Layout(width='500px'), 
)
 
but_allregions = widgets.Button(description = 'Select all regions')
but_allregions_toggle = True
def select_allregions(*args):
    global but_allregions_toggle
    if but_allregions_toggle:
        regions.value = tuple(regions.options)
        but_allregions.description = 'Unselect all regions'
        but_allregions_toggle = False
    else:
        regions.value = []
        but_allregions.description = 'Select all regions'
        but_allregions_toggle = True
but_allregions.on_click(select_allregions)
 
num_pairs_scope = widgets.ToggleButtons(
    options=['Overall', 'Per sector'],
    description='The number of pairs will be defined by:',
    button_style='',
    tooltips=['A given number of pairs will be selected', 'A given number of pairs per sector will be selected']
)
 
num_pairs_value = widgets.SelectionSlider(
    options=[10, 20, 50, 150, 200, 300],
    value=50,
    description='Number of pairs:',
    style={'description_width': 'auto'},
    layout=widgets.Layout(width='500px'),
    continuous_update=False
)
 
quantile = widgets.FloatSlider(
    value=0.50,
    min=0.05,
    max=1.00,
    step=0.05,
    description=f'Quantile for pair selection**',
    orientation='horizontal',
    readout=True,
    readout_format='.2f',
    style={'description_width': 'auto'},
    layout=widgets.Layout(width='500px'),
)
quantile_info = widgets.Label('**By selecting X above, the 1-X pairs with the highest coefficients will be excluded',layout=widgets.Layout(width='500px'), )
 
run_button = widgets.Button(description="Generate Pairs", disabled=True)

def fill_form_holder(form_holder):
    """
    Fills the form holder with the values from the widgets
    """
    form_holder[FormTag.UPDATE_FREQ] = update_freq.value
    form_holder[FormTag.PC_WINDOW] = pc_window.value
    form_holder[FormTag.PC_CORRELWEIGHT] = pc_correlweight.value
    form_holder[FormTag.REPEAT_COMPANIES] = repeat_companies.value
    form_holder[FormTag.SECTORS] = sectors.value
    form_holder[FormTag.REGIONS] = regions.value
    form_holder[FormTag.PAIRS_SCOPE] = num_pairs_scope.value
    form_holder[FormTag.NUM_PAIRS_VALUE] = num_pairs_value.value
    form_holder[FormTag.QUANTILE] = quantile.value
    
    return form_holder

def clear_form_holder(form_holder):
    """
    Clears the form holder
    """
    form_holder[FormTag.UPDATE_FREQ] = None
    form_holder[FormTag.PC_WINDOW] = None
    form_holder[FormTag.PC_CORRELWEIGHT] = None
    form_holder[FormTag.REPEAT_COMPANIES] = None
    form_holder[FormTag.SECTORS] = None
    form_holder[FormTag.REGIONS] = None
    form_holder[FormTag.PAIRS_SCOPE] = None
    form_holder[FormTag.NUM_PAIRS_VALUE] = None
    form_holder[FormTag.QUANTILE] = None
    return form_holder

def get_the_pairs(form_holder):
    """Pair queries and scoring are not connected to the interface yet."""
    raise NotImplementedError("Pair generation is unfinished.")

def generate_pairs(form_holder):
    """Collect the widget settings for pair generation."""
    clear_output(wait=True)
    
    with output:
        clear_output(wait=True)
        clear_form_holder(form_holder)
        form_holder = fill_form_holder(form_holder)
        
        return get_the_pairs(form_holder)
        
 
run_button.on_click(lambda btn: generate_pairs(Form_Holder))
 
display(update_freq, pc_window, pc_correlweight, pc_correlweight_info,
        repeat_companies, sectors, but_allsectors, regions, but_allregions, num_pairs_scope, num_pairs_value, quantile, quantile_info, run_button, output)
 
 
```
